```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data = df = pd.read_csv('GlobalLandTemperaturesByMajorCity.csv')
data.head()
```

```python
stolica_kraju = df[df['City'] == 'Berlin']
```

```python
print(stolica_kraju)
```

W przypadku pliku MajorCity.csv jest tylko kilka miast (brak Warszawy).

```python
stolica_kraju['dt'] = pd.to_datetime(stolica_kraju['dt'])
stolica_kraju = stolica_kraju[(stolica_kraju['dt'].dt.year >= 1970) & (stolica_kraju['dt'].dt.year <= 2020)]
stolica_kraju = stolica_kraju[stolica_kraju['AverageTemperature'].notna()]
print(stolica_kraju)
```

Dane z zakresu 1970 do 2020, wartosci Nan nie zostaly wyswietlone

```python
stolica_kraju['year'] = stolica_kraju['dt'].dt.year
srednia_temp = stolica_kraju.groupby('year')['AverageTemperature'].mean().reset_index()
print(srednia_temp)
```

plt.figure(figsize=(10, 5))
sns.lineplot(data='srednia_temp', x='year', y='AverageTemperature')
plt.title('Srednia roczna temperatura')
plt.xlabel('Rok')
plt.ylabel('Temperatura [*C]')
plt.show()

Oblicz średnią roczną temperaturę i narysuj wykres liniowy zmian temperatur w czasie. -Oblicz różnicę między temperaturą średnią z lat 1970–1990 a 2000–2020. -Wyświetl wykres z linią trendu (użyj sns.regplot lub sns.lineplot).

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.lineplot(data=srednia_temp, x='year', y='AverageTemperature')
plt.title('Srednia roczna temperatura dla stolica_kraju')
plt.xlabel('Rok')
plt.ylabel('Temperatura')
plt.show()
```
