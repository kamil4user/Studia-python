```python
import pandas as pd
#baza coffe shop sales
df = pd.read_csv('CoffeeShopSales.csv', sep=';')
df.head()
```


2. Sprawdź zużycie pamięci:

python

df.info(memory_usage="deep")

df.memory_usage(deep=True)

· info(memory_usage="deep") – pokazuje łączną pamięć DataFrame (z uwzględnieniem kolumn typu object).

· memory_usage(deep=True) – pokazuje pamięć dla każdej kolumny; na tej podstawie widzisz, co „najbardziej boli”.

```python
df.info(memory_usage="deep")
df.memory_usage(deep=True)
```

Najwięcej pamięci zużywają kolumny typu object.
Najwięcej pamięci zużywa kolumna product_detail.
Stringów (object) jest więcej, dlatego pożerają więcej pamięci niż int-y.
