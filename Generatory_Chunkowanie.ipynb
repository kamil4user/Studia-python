4.3 Prosty generator.

```python
def czytaj_linie(filename):
    with open(filename, "r", encoding="utf-8") as f:
        for line in f:
            yield line.strip()
```

· Generator trzyma w pamięci tylko jedną linię na raz, a nie cały plik.
